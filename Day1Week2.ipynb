````python
reply = response_3['content']

# Tạo display ban đầu
display_handle = display(Markdown(""), display_id=True)

# Mô phỏng streaming: chia reply theo từng từ và cập nhật dần
partial_reply = ""
for word in reply.split(" "):
    partial_reply += word + " "
    # Loại bỏ các ký tự không cần thiết nếu cần (giống như trong mã gốc)
    formatted_reply = partial_reply.replace("```", "").replace("markdown", "")
    update_display(Markdown(formatted_reply), display_id=display_handle.display_id)
    time.sleep(0.2)  # giả lập độ trễ (0.2 giây mỗi từ)

print("Number of words:", len(reply.split(" ")))
````
